```python
import pandas as pd
from zipfile import ZipFile 
import os
```

```python
zip_files_directory = '../..'
extract_to_directory = '../data/raw'
```

```python
zipfiles = []
for i in os.listdir(zip_files_directory):
    if i.endswith(".zip"):
        zipfiles.append(i)

zipfiles
```

```python
for i in zipfiles:
    with ZipFile(f"{zip_files_directory}\\{i}") as ZipObject:
        ZipObject.extractall(path=f"{extract_to_directory}//{i}")
```

```python
folders = ["raw_tennis_power_parquet",
           "raw_odds_parquet",
           "raw_point_by_point_parquet",
           "raw_votes_parquet",
           "raw_match_parquet",
           "raw_statistics_parquet"]
folders
```

```python
dataframes = {
    "MatchEventInfo": pd.DataFrame(),
    "PeriodInfo": pd.DataFrame(),
    "MatchVotesInfo": pd.DataFrame(),
    "MatchTournamentInfo": pd.DataFrame(),
    "MatchSeasonInfo": pd.DataFrame(),
    "MatchRoundInfo": pd.DataFrame(),
    "MatchVenueInfo": pd.DataFrame(),
    "MatchHomeTeamInfo": pd.DataFrame(),
    "MatchAwayTeamInfo": pd.DataFrame(),
    "MatchHomeScoreInfo": pd.DataFrame(),
    "MatchAwayScoreInfo": pd.DataFrame(),
    "MatchTimeInfo": pd.DataFrame(),
    "GameInfo": pd.DataFrame(),
    "OddsInfo": pd.DataFrame(),
    "PowerInfo": pd.DataFrame(),
}

for zip_folder in zipfiles:
    for folder in folders:
        folder_path = f"..\\data\\raw\\{zip_folder}\\data\\raw\\{folder}"
        
        for file in os.listdir(folder_path):
            df = pd.read_parquet(f"{folder_path}\\{file}", engine='pyarrow')

            if df.empty:
                continue

            if file.startswith("event"):
                key = "MatchEventInfo"
            elif file.startswith("statistics"):
                key = "PeriodInfo"
            elif file.startswith("votes"):
                key = "MatchVotesInfo"
            elif file.startswith("tournament"):
                key = "MatchTournamentInfo"
            elif file.startswith("season"):
                key = "MatchSeasonInfo"
            elif file.startswith("round"):
                key = "MatchRoundInfo"
            elif file.startswith("venue"):
                key = "MatchVenueInfo"
            elif file.startswith("home_team_1"):
                key = "MatchHomeTeamInfo"
            elif file.startswith("away_team_1"):
                key = "MatchAwayTeamInfo"
            elif file.startswith("home_team_score"):
                key = "MatchHomeScoreInfo"
            elif file.startswith("away_team_score"):
                key = "MatchAwayScoreInfo"
            elif file.startswith("time"):
                key = "MatchTimeInfo"
            elif file.startswith("pbp"):
                key = "GameInfo"
            elif file.startswith("odds"):
                key = "OddsInfo"
            elif file.startswith("power"):
                key = "PowerInfo"
            else:
                continue

            df = df.dropna(axis=1, how="all")

            if not df.empty:
                dataframes[key] = pd.concat([dataframes[key], df], ignore_index=True, sort=False)
```

```python
output_directory = r'..\data\processed'

for name, df in dataframes.items():
        df.to_csv(f"{output_directory}\\{name}.csv", index=True)
```
